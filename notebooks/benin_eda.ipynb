```python
# Select only numeric columns
numeric_cols = df.select_dtypes(include='number').columns

# Fill missing values in numeric columns with their medians
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].median())
```

```python
# CELL 5: cleaning pipeline (example)
df_clean = df.copy()

# drop rows where more than half of key columns are missing
key_cols = [c for c in ["GHI","DNI","DHI","ModA","ModB","Tamb"] if c in df_clean.columns]
df_clean["num_key_missing"] = df_clean[key_cols].isna().sum(axis=1)
initial_len = len(df_clean)
df_clean = df_clean[df_clean["num_key_missing"] <= (len(key_cols) // 2)].drop(columns=["num_key_missing"])
dropped_rows = initial_len - len(df_clean)
print(f"Dropped {dropped_rows} rows due to >50% missing key columns")

# Impute median for remaining missing in key columns
for col in key_cols:
    if df_clean[col].isna().sum() > 0:
        median_val = df_clean[col].median()
        df_clean[col].fillna(median_val, inplace=True)
        print(f"Imputed median for {col}: {median_val} (count={df[col].isna().sum()})")

# Clip physically impossible values (simple sanity rules)
if "GHI" in df_clean.columns:
    df_clean["GHI"] = df_clean["GHI"].clip(lower=0)  # GHI can't be negative
if "WS" in df_clean.columns:
    df_clean["WS"] = df_clean["WS"].clip(lower=0)

# Optionally remove extreme outliers flagged by Z but not too many rows:
z = np.abs(stats.zscore(df_clean[num_cols].dropna()))
z_df = pd.DataFrame(z, index=df_clean[num_cols].dropna().index, columns=num_cols)
outlier_idx = z_df[(z_df > 4).any(axis=1)].index  # stricter cutoff for removal
print("Rows to remove by strict z>4:", len(outlier_idx))
df_clean = df_clean.drop(index=outlier_idx)

# Final size
print("Final shape:", df_clean.shape)
```
